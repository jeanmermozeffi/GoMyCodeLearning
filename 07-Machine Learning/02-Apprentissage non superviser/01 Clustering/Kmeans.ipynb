## Exemple de Code Elbow Method

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
```

```python
# Générer un jeu de données factice pour l'exemple
# X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)
```

```python
# Liste pour stocker les sommes des erreurs quadratiques (SSE) pour différents k
sse = []

# Tester plusieurs valeurs de k (par exemple de 1 à 10)
k_range = range(1, 11)

for k in k_range:
    # Appliquer K-Means avec k clusters
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)

    # Calculer la somme des erreurs quadratiques (SSE)
    sse.append(kmeans.inertia_)
```

```python
# Tracer la courbe SSE en fonction de k
plt.figure(figsize=(8, 5))
plt.plot(k_range, sse, marker='o')
plt.title('Méthode du Coude')
plt.xlabel('Nombre de clusters (k)')
plt.ylabel('Somme des erreurs quadratiques (SSE)')
plt.show()
```
