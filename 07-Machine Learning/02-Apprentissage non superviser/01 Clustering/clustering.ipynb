## chargement de données

```python
import pandas as pd
```

```python
df = pd.read_csv('../../../DataSets/07/Mall_Customers.csv')
df.head(10)
```

```python
# Sélection des colonnes pertinentes pour notre clustering
X = df[['Age', 'Gender', 'Annual Income (k$)', 'Spending Score (1-100)']]
X.values
```

```python
## Encodage des valeurs categorielles
X.loc[:, "Gender"] = X.loc[:, "Gender"].map({'Male': 1, 'Female': 0})
X
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
# Normalisations des données
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled
```

```python
# Appliquer le clustering agglomeratif
from sklearn.cluster import AgglomerativeClustering

model = AgglomerativeClustering(n_clusters=5, metric='euclidean', linkage='complete')
clusters_label = model.fit_predict(X)
```

```python
# Creation dun Dataframe pour les label de  nos clusters
agglomerative = pd.DataFrame(clusters_label)
agglomerative[:5]
```

```python
import matplotlib.pyplot as plt
```

```python
df.columns
```

## Non Normalise

```python
# Visualisation des clusters en 2D
plt.figure(figsize=(10, 6))
plt.scatter(X['Annual Income (k$)'], X['Spending Score (1-100)'], c=clusters_label, cmap='rainbow', s=50)
plt.xlabel('Annual Income (k$) (Normalisé)')
plt.ylabel('Spending Score (1-100)')
plt.title('Clusters des clients (AgglomerativeClustering)')
plt.show()
```

## Interprétation des groupes agglomératifs
Les clients en vert : 
- ont un petit revenu annuel mais ont tendance à dépenser beaucoup. 
- On peut les appeler les extravagants.
- Les clients en jaune : ce sont des clients à faibles revenus et qui ne dépensent pas beaucoup. On peut les appeler les prudents.
- Clients en bleu : Clients avec un revenu moyen et des dépenses moyennes. Ce sont des clients normaux.
- Clients en bleu clair : clients avec un revenu annuel élevé et un score de dépenses élevé. Ce sont des clients généreux.
- Les clients en violet : ce sont des clients qui ont un revenu annuel élevé mais qui ne dépensent pas beaucoup. On peut les appeler les moins chers.

# Visualisation du dendrogramme

```python
from scipy.cluster.hierarchy import dendrogram, linkage
import scipy.cluster.hierarchy as shc
```

```python
plt.figure(figsize=(10, 6))
linked = linkage(X_scaled, 'complete')
dendrogram(linked, truncate_mode='lastp', p=12, show_contracted=True)
plt.xlabel('Annual Income (k$) (Normalisé)')
plt.ylabel('Spending Score (1-100)')
plt.title('Clusters des clients (AgglomerativeClustering)')
plt.show()
```

## Normalisation

```python
model_scaler = AgglomerativeClustering(n_clusters=5, metric='euclidean', linkage='complete')
clusters_label_scaler = model_scaler.fit_predict(X_scaled)
```

```python
# Visualisation des clusters en 2D
plt.figure(figsize=(10, 6))
plt.scatter(X_scaled[:, 2], X_scaled[:, 3], c=clusters_label_scaler, cmap='rainbow')
plt.xlabel('Annual Income (k$) (Normalisé)')
plt.ylabel('Spending Score (1-100)')
plt.title('Clusters des clients (AgglomerativeClustering)')
plt.show()
```
