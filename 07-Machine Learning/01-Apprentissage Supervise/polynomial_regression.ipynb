```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
jours = np.linspace(start=1, stop=100, num=100).reshape(-1, 1)
hauteurs = 2 + 0.5 * jours - 0.01 * jours**2 + np.random.randn(100, 1)
```

### Visualisation des Données

```python
plt.scatter(jours, hauteurs, color='black', label='Données réelles')
plt.xlabel('Temps (jours)')
plt.ylabel('Hauteur de la plante (cm)')
plt.title('Croissance des plantes')
plt.legend()
plt.show()
```

### 2. Analyse des Résidus

```python
from sklearn.linear_model import LinearRegression
```

```python
lig_model = LinearRegression()
lig_model.fit(hauteurs, jours)
y_preds = lig_model.predict(hauteurs)
```

```python
# Calcul des residus
residuals = hauteurs - y_preds
```

```python
# Visualisation des résidus
plt.scatter(jours, residuals, color='red')
plt.xlabel('Variable indépendante (X)')
plt.ylabel('Résidus')
plt.title('Analyse des résidus')
plt.show()
```

```python
# Transformation des caractéristiques en polynômes de degré 2
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree=2)
jours_poly = poly.fit_transform(jours)
```

```python
# Ajustement du modèle de régression polynomiale
poly_reg = LinearRegression()
poly_reg.fit(jours_poly, hauteurs)
y_poly_preds = poly_reg.predict(jours_poly)
```

```python
# Comparaison des modèles
from sklearn.metrics import mean_squared_error, r2_score

lin_mse = mean_squared_error(hauteurs, y_preds)
poly_mse = mean_squared_error(hauteurs, y_poly_preds)
lin_r2 = r2_score(hauteurs, y_preds)
poly_r2 = r2_score(hauteurs, y_poly_preds)

print("Régression linéaire - Mean Squared Error:", lin_mse)
print("Régression polynomiale - Mean Squared Error:", poly_mse)
print("Régression linéaire - R^2 Score:", lin_r2)
print("Régression polynomiale - R^2 Score:", poly_r2)
```

```python
# Visualisation des résultats
plt.scatter(jours, hauteurs, color='black', label='Données réelles')
plt.plot(hauteurs, y_preds, color='red', linewidth=2, label='Régression linéaire')
plt.plot(hauteurs, y_poly_preds, color='blue', linewidth=2, label='Régression polynomiale')
plt.xlabel('Variable indépendante (X)')
plt.ylabel('Variable dépendante (y)')
plt.title('Comparaison des modèles')
plt.legend()
plt.show()
```

```python
# Calcul des residus
residuals_poly = hauteurs - y_poly_preds

plt.scatter(jours, residuals_poly, color='yellow')
plt.axhline(y=0, color='red', linestyle='-')
plt.xlabel('Variable indépendante (X)')
plt.ylabel('Résidus')
plt.title('Analyse des résidus')
plt.show()
```
