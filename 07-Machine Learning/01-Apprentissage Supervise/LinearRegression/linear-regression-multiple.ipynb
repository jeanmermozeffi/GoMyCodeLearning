### Paramettres

```python
df = pd.read_csv(path_file)
df.head()
```

## Feature Selection

```python
df.columns
```

```python
cols = [
    'identifiant',
    'date',
    'prix',
    'chambres',
    'salles de bains',
    'surface habitable',
    'surface du terrain',
    'étages',
    'bord de eau',
    'vue',
    'état',
    'qualité',
    'surface au-dessus du sol',
    'surface du sous-sol',
    'année de construction',
    'année de rénovation',
    'code postal',
    'latitude',
    'longitude',
    'surface habitable moyenne des 15 maisons les plus proches',
    'surface du terrain moyenne des 15 maisons les plus proches'
]
```

```python
features = ["bedrooms", "bathrooms", "sqft_living", "sqft_above", "floors", "waterfront", "view", "condition", "grade",
            "sqft_living15", "sqft_lot15", "yr_built", "yr_renovated"]

target = "price"
```

### Division des : Cross Validation

```python
from sklearn.model_selection import train_test_split
```

```python
y = df[target]
X = df[features]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
print("Taille train: ", X_train.shape, y_train.shape)
print("Taille test: ", X_test.shape, y_test.shape)
```

### Visualisation des données d'entraînement et de test

```python
plt.figure(figsize=(10, 6))

# Visualisation des données d'entraînement
plt.scatter(X_train["sqft_living"], y_train, color='blue', label='Données d\'entraînement')

# Visualisation des données de test
plt.scatter(X_test["sqft_living"], y_test, color='green', label='Données de test')

plt.xlabel('Surface habitable (sqft_living)')
plt.ylabel('Prix (price)')
plt.title('Visualisation des données d\'entraînement et de test')
plt.legend()
plt.show()
```

## Model

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()

model.fit(X_train, y_train)
y_preds = model.predict(X_test)
```

```python
y_preds[:5]
```

```python
y_train[:5]
```

## Evaluation Metrics

```python
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)
```

```python
r2 = r2_score(y_test, y_preds)
rmse = np.sqrt(mean_squared_error(y_test, y_preds))
mae = mean_absolute_error(y_test, y_preds)

print("R squared: {0}".format(r2))
print("RMSE: {0}".format(rmse))
print("MAE: {0}".format(mae))
```
