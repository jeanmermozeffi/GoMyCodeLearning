### Implémentation de la Fonction Sigmoïde

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Définition de la fonction sigmoïde
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
z_train = np.linspace(-10, 10, 100)
z_train
```

```python
sigmoid_values = sigmoid(z_train)
sigmoid_values
```

```python
# Affichage graphique
plt.plot(z_train, sigmoid_values)
plt.xlabel('z')
plt.ylabel('Sigmoid(z)')
plt.title('Fonction Sigmoïde')
plt.grid(True)
plt.show()
```

## Mesures de Performance

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
```

```python
y_true = np.random.choice([0, 1], size=100)
y_pred = np.random.choice([0, 1], size=100)
```

```python
y_true
```

```python
y_pred
```

```python
# Calcul de la matrice
cm = confusion_matrix(y_true, y_pred)
```

```python
# Afficher la matrice
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Positif', 'Négative'])
disp.plot(cmap=plt.cm.Blues)
plt.title('Matrice de confusion')
plt.show()                          
```
