## Importation des bibliotheque

```python
from sklearn import datasets #Pour des données de Test
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report
```

## Chargement des données

```python
iris = datasets.load_iris()
X = iris.data # Caractéristique (Features)
y = iris.target # Étiquette (label)

# Diviser la donnée en Train, Test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
X_train.shape , X_test.shape , y_train.shape , y_test.shape
```

## Créer un modèle SVM

```python
# Un modele SVM avec un noyau linéaire
model = SVC(kernel='linear')

# Entrainement du modele
model.fit(X_train, y_train)

# Faire des prédictions sur l'ensemble des données de test
y_pred = model.predict(X_test)
```

## Evaluer les performances 

```python
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy : {accuracy * 100:.2f}")
```

```python
print(classification_report(y_test, y_pred))
```
