```python
# Importation des bibliothèques nécessaires
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

from sklearn.model_selection import train_test_split
from  sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
iris = load_iris()
X = iris.data[:, :4]
y = iris.target
```

## Avec les données de la fonction load_iris() la variable cible est déjà encode

```python
# Encoder les étiquettes
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)
```

```python
y_encoded
```

```python
# Division des données en ensembles d'entraînement et de test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=30)
```

```python
# Normalisation des données
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
X_train
```

```python
# Scatter plot des données d'entraînement
plt.figure(figsize=(10, 6))
plt.scatter(X_train[:, 0], X_train[:, 1], c=y_train, cmap='viridis', marker='o', edgecolor='k', s=100)
plt.title('Scatter plot des données d\'entraînement')
plt.xlabel('Longueur des sépales (standardisée)')
plt.ylabel('Largeur des sépales (standardisée)')
plt.show()
```

```python
# Création et entraînement du modèle KNN
k = 5
knn = KNeighborsClassifier(n_neighbors=k)
knn.fit(X_train, y_train)
```

```python
# Prédiction sur l'ensemble de test
y_pred = knn.predict(X_test)
```

```python
# Évaluation du modèle
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
class_report = classification_report(y_test, y_pred)
```

```python
print(f'Accuracy: {accuracy}')
```

```python
print('Confusion Matrix:')
print(conf_matrix)
```

```python
print('Classification Report:')
print(class_report)
```

```python
# Matrice de Confusion
plt.figure(figsize=(8, 6))
plt.imshow(conf_matrix, interpolation='nearest', cmap='Blues')
plt.title("Confusion Matrix")
plt.colorbar()
tick_marks = np.arange(len(iris.target_names))
plt.xticks(tick_marks, iris.target_names, rotation=45)
plt.yticks(tick_marks, iris.target_names)
plt.xlabel('Classe prédite')
plt.ylabel('Classe réelle')

for i, in range(len(iris.target_names)):
    for j in range(len(iris.target_names)):
        plt.text(j, i, conf_matrix[i, j], horizontalalignment='center', color='black')

plt.show()
```
